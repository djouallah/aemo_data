```python
ws                    = 'duckrun'
lh                    = 'data'
schema                = 'aemo'
business_logic        = f"abfss://{ws}@onelake.dfs.fabric.microsoft.com/{lh}.Lakehouse/Files/transformation"
bim_url               = "/fabric_demo/semantic_model/directlake.bim"
```

```python

import duckrun
con = duckrun.connect(f"{ws}/{lh}.lakehouse/{schema}")
```

```python
#create lakehouse if not exists
con.create_lakehouse_if_not_exists(lh)
```

```python
con.copy('/fabric_demo/transformation','transformation',['.py','.sql'],overwrite=True)
```

```python
con.deploy_notebook(
    '/fabric_demo/electricity.ipynb',
    overwrite=True,
    parameters={
        "ws": ws,
        "lh": lh,
        "schema": schema,
        "business_logic": business_logic ,
        "nbr_days_download": 288
    }
)
```

```python
con.schedule_notebook('electricity', schedule_type='interval', interval_minutes=60)
```

```python
%%time
con.deploy_bim(bim_url)
```
