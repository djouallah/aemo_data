```python
import pandas
from typing import Iterable
import json
import time
import sempy.fabric as fabric
tom = fabric.create_tom_server() #get CLR loaded
from datetime import datetime
import requests


from Microsoft.AnalysisServices.AdomdClient import AdomdConnection

def run_query(con: AdomdConnection, query:str) -> int:
    cmd = con.CreateCommand()
    cmd.CommandText = query
    rows = 0
    rdr = cmd.ExecuteReader()
    while rdr.Read():
        rows=rows+1
    rdr.Close()
    return rows


def load_queries(fn: str) -> list:
    queries = []

    # Load JSON string either from local file or GitHub URL
    if fn.startswith("http://") or fn.startswith("https://"):
        resp = requests.get(fn)
        resp.raise_for_status()
        # ensure BOM-safe decoding
        json_string = resp.content.decode("utf-8-sig")
    else:
        with open(fn, encoding="utf-8-sig") as f:
            json_string = f.read()

    # Parse JSON
    d = json.loads(json_string)

    visual_name = ""
    query_text = ""

    for e in d.get("events", []):
        if e.get("name") == "Visual Container Lifecycle":
            visual_name = e["metrics"].get("visualTitle", "")

        if e.get("name") == "Execute DAX Query":
            query_text = e["metrics"].get("QueryText", "")
            queries.append({
                "visual_name": visual_name,
                "query_text": query_text
            })

    return queries


def run_perf_scenario(con: AdomdConnection, queries: Iterable[str], i: int) -> list:
    results = []
    for qn, q in enumerate(queries, start=1):
        start = time.time()
        query_text = q["query_text"]
        visual_name = q["visual_name"]
        rows = run_query(con, query_text)
        duration = time.time() - start
        
        result = {
            "loadtest_id": loadtestId,
            "model": model,
            "concurrent_threads": concurrent_threads,
            "iterations": iterations,
            "delay_sec": delay_sec,
            "query_number": qn,
            "iteration": i,
            "rows": rows,
            "duration": duration,
            "start_time": start,
            "start_time_dt": datetime.fromtimestamp(start),
            "thread_id": threadId,
            "cache" : cache,
            "totalrows" : totalrows,
            "nbr_queries" : nbr_queries
        }
        results.append(result)
        time.sleep(delay_sec)
    return results
```

```python
import sempy.fabric as fabric
import pandas as pd
import uuid
import time
from Microsoft.AnalysisServices.AdomdClient import AdomdConnection, AdomdCommand
import deltalake
from deltalake.writer import write_deltalake

tom = fabric.create_tom_server()  # get CLR loaded

token = notebookutils.credentials.getToken("pbi")

if xmla_endpoint is None:
    xmla_endpoint = f"powerbi://api.powerbi.com/v1.0/myorg/{notebookutils.runtime.context['currentWorkspaceName']}"

constr = f"Data Source={xmla_endpoint};Initial Catalog={model};password={token};Timeout=7200;"
if effective_username is not None:
    constr += f"EffectiveUserName={effective_username};"
if customdata is not None:
    constr += f"CustomData={customdata};"
if roles is not None:
    constr += f"Roles={roles};"

print(constr)
con = AdomdConnection(constr)
try:
    con.Open()
    
    # Clear cache before running queries
    print("Clearing cache...")
    
    # Get the dataset ID from the model name
    workspace_name = notebookutils.runtime.context['currentWorkspaceName']
    datasets = fabric.list_datasets(workspace=workspace_name)
    model_id = datasets[datasets['Dataset Name'] == model]['Dataset ID'].values[0]
    
    clear_cache_xmla = f"""
                        <ClearCache xmlns="http://schemas.microsoft.com/analysisservices/2003/engine">
                        <Object>
                            <DatabaseID>{model_id}</DatabaseID>
                        </Object>
                        </ClearCache>
                        """
    
    fabric.execute_xmla(dataset=model_id, xmla_command=clear_cache_xmla, workspace=workspace_name)
    print("Cache cleared successfully")
    
    queries = load_queries(perf_analyzer_filename)[:nbr_queries]
    all_results = []
    print(f"starting {iterations} iterations with {delay_sec} think time between")
    for i in range(iterations):
        results: list = run_perf_scenario(con, queries, i)
        if i < iterations:
            print(f"Completed iteration, starting {delay_sec}sec think time")
            time.sleep(delay_sec)
        all_results += results
    con.Close()
    MAX_RETRIES = 5
    RETRY_DELAY_SEC = 2
    df = pd.DataFrame(all_results)
    attempt = 0
    while attempt < MAX_RETRIES:
        try:
            write_deltalake(delta_path, df, mode="append")
            print(f"Delta write successful on attempt {attempt+1}")
            break
        except Exception as write_err:
            attempt += 1
            print(f"Delta write failed on attempt {attempt}: {write_err}")
            if attempt < MAX_RETRIES:
                print(f"Retrying in {RETRY_DELAY_SEC} seconds...")
                time.sleep(RETRY_DELAY_SEC)
            else:
                print("Max retries reached. Failing.")
                raise write_err
except Exception as e:
    print(e)
    raise
```
